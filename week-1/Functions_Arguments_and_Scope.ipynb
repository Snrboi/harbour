## 1. What is a function?

A **function** is a reusable block of code that performs a particular task. It helps us avoid repeating code and gives each task a clear name.

Think of a function as a small machine: it may take **inputs**, perform **work**, and produce an **output**.

- `def` begins a function definition.
- The indented lines form the function body.
- Defining a function does not run it. Calling the function does.

````python
def greet():
    print("Welcome to HARBOUR!")

greet()  # Call the function
````

**Ask the group:** What would happen if we defined `greet()` but never called it?

**Answer:** The function would be defined, but its body would not run.

## 2. Parameters and arguments

A **parameter** is a named input slot in the function definition. An **argument** is the actual value passed when the function is called.

Memory trick: parameter = slot; argument = value placed in the slot.

````python
def greet_person(name):  # name is a parameter
    print(f"Hello, {name}!")

greet_person("Ada")   # "Ada" is an argument
greet_person("Tega")  # "Tega" is another argument
````

### A function with two parameters

Python matches positional arguments to parameters from left to right.

````python
def calculate_total(price, quantity):
    total = price * quantity
    return total

result = calculate_total(1500, 3)
print(result)
````

In `calculate_total(1500, 3)`, the arguments are `1500` and `3`. Python assigns them to `price` and `quantity` respectively.

**HARBOUR connection:** A function could calculate the total cost of a customer's order without needing to know the whole application.

## 3. `print()` versus `return`

- `print(value)` displays information for a person to see.
- `return value` sends a result back to the code that called the function.
- A function without an explicit `return` returns `None` by default.

````python
def add_and_print(a, b):
    print(a + b)

def add_and_return(a, b):
    return a + b

shown = add_and_print(2, 3)
saved = add_and_return(2, 3)

print("Value from print-only function:", shown)
print("Value returned by function:", saved)
print("Returned value can be reused:", saved * 10)
````

**Explain the output:** Both functions display or produce the number `5`, but only the second returns it. Therefore `shown` is `None`, while `saved` is `5` and can be used in another calculation.

**Rule of thumb:** Use `return` when other code needs the result; use `print` when you want to display it.

## 4. Positional, keyword, and default arguments

### Positional arguments
Python matches the values to parameters by their position.

````python
def describe_order(item, quantity):
    return f"{quantity} x {item}"

print(describe_order("notebook", 4))
````

### Keyword arguments
Name the parameter in the call. This makes the code easier to read and lets you supply arguments in a different order.

````python
print(describe_order(quantity=4, item="notebook"))
````

### Default parameter values
A parameter can have a default. If the caller leaves it out, Python uses that default.

````python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Ada"))
print(greet("Tega", greeting="Good morning"))
````

Here `greeting` has the default value `"Hello"`. The first call uses it; the second overrides it.

**Common rule:** Required parameters should come before parameters with default values in a function definition.

## 5. Scope: where can a variable be used?

**Scope** is the part of a program where a name is accessible.

### Local scope
A variable created inside a function is usually local to that function. It cannot normally be accessed directly outside it.

````python
def make_message():
    message = "Stock check complete"  # local variable
    print(message)

make_message()
# Uncomment the next line to see the NameError:
# print(message)
````

The name `message` is local to `make_message()`. If you uncomment `print(message)` outside the function, Python raises `NameError` because that name is not available in the outer scope.

Function parameters are local too.

````python
def double(number):
    result = number * 2
    return result

print(double(6))
# `number` and `result` are local to double().
# print(number)  # This would raise NameError
````

### Global scope
A name created outside functions is in the module's global scope. A function can read it, but depending on global variables—especially changing them—can make code harder to understand.

````python
business_name = "Harbour Supplies"

def print_business():
    print(f"Welcome to {business_name}")

print_business()
````

### Prefer explicit inputs

Instead of relying on a global tax rate inside a function, pass the rate in as an argument. That makes the function easier to reuse and test.

````python
def add_tax(price, rate):
    return price + (price * rate)

tax_rate = 0.075
print(add_tax(1000, tax_rate))
print(add_tax(1000, 0.10))  # Try a different rate
````

## 6. Practical HARBOUR example: stock status

This function uses parameters, a local variable, conditionals, and `return`.

````python
def stock_status(item_name, quantity):
    if quantity <= 0:
        status = "OUT OF STOCK"
    elif quantity < 5:
        status = "LOW STOCK"
    else:
        status = "IN STOCK"

    return f"{item_name}: {status} (quantity: {quantity})"

print(stock_status("Hand sanitizer", 0))
print(stock_status("Gloves", 3))
print(stock_status("Notebooks", 24))
````

**Walk through the code:**
1. `item_name` and `quantity` are parameters.
2. Each function call supplies arguments.
3. `status` is a local variable.
4. The conditional chooses a status.
5. `return` sends the formatted message back.
6. `print()` displays the returned message.

This function keeps its inputs and output clear and does not change a global variable.

## 10. One-minute teach-back script

> A function is a reusable block of code defined with `def` and executed when called. Parameters are the named input slots in the definition, while arguments are the actual values passed during a call. `return` sends a result back to the caller, while `print` displays information. Scope describes where a variable name can be accessed. Variables created inside a function are usually local to that function, while names created outside functions are in global scope. In HARBOUR, functions let us separate tasks—such as checking stock status—into small, reusable, testable pieces. Passing inputs through parameters and returning results makes those pieces easier to maintain.

## Final recap
- **Define:** `def function_name(parameters):`
- **Call:** `function_name(arguments)`
- **Parameter:** named input in the definition.
- **Argument:** value passed into a call.
- **Return:** sends a result back.
- **Scope:** controls where a name can be accessed.
- Prefer clear inputs and return values over unnecessary global-state changes.
