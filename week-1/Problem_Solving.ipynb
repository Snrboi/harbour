## A practical team workbook

**Goal:** Take one problem and produce a clear solution with evidence that it works.

Use five steps:
1. **Restate:** Say exactly what the solution must do.
2. **Examples:** Write specific inputs and correct outputs.
3. **Plan:** Break the work into small actions, then carry them out.
4. **Test:** Compare your results with the expected results.
5. **Reflect:** Record what changed and what you learned.

You will work through a buying decision, a delivery calculator, and an article-grouping function. Then you will solve a problem without a supplied solution.

**To run code:** Open this file in Jupyter, VS Code, or Google Colab. Run code cells from top to bottom. Only Python 3 is required.

## 1. The method to use for any exercise

| Step | Write this | You are ready to move on when… |
|---|---|---|
| Restate | “Given ___, produce ___, following these rules: ___.” | Someone else could tell a correct answer from an incorrect one. |
| Examples | At least one ordinary case and one case at a limit or exception. Write the expected result for each. | You can explain each expected result using the rules. |
| Plan | Numbered actions. Say what each action reads, calculates, stores, or returns. | You can follow the actions by hand on one example. |
| Test | Expected result, actual result, and pass/fail. | You have compared actual results with your examples. Fix failures and rerun. |
| Reflect | One specific lesson and one next action. | You can name what worked, failed, or remains untested. |

### When you get stuck
- **The request is unclear:** Ask a question. If you must assume something, write the assumption.
- **You cannot make a plan:** Solve one small example by hand. Write down the actions you took.
- **A step says “process the data”:** Replace it with a specific action, such as “convert tags to a `frozenset`.”
- **A test fails:** Find the first step where actual and expected results differ. Change that step and rerun all tests.
- **No solution meets the rules:** Explain which rules conflict. Do not silently remove a rule.

**Examples come before the solution. Tests check the solution afterward.** They can use the same inputs.

## 3. Worked log: delivery-fee calculator

### 1 — Restate
Write `delivery_fee(distance_km)` using these fictional rules:
- 0 through 5 km → ₦1,000.
- Above 5 through 10 km → ₦1,500.
- Above 10 through 20 km → ₦2,500.
- Above 20 km → `None`, meaning delivery is unavailable.
- Negative distance → raise `ValueError`.

Assume the input is a finite number. No rounding or extra fees apply.

### 2 — Examples
| Input | Expected | Rule checked |
|---:|---|---|
| 3 | 1000 | Ordinary distance |
| 5 | 1000 | Exact limit |
| 5.1 | 1500 | Just above the limit |
| 10 | 1500 | Exact limit |
| 10.1 | 2500 | Just above the limit |
| 20 | 2500 | Last served distance |
| 20.1 | None | Outside the service area |
| -1 | ValueError | Invalid distance |

### 3 — Plan
1. Reject a negative distance.
2. If the distance is at most 5, return 1000.
3. Otherwise, if it is at most 10, return 1500.
4. Otherwise, if it is at most 20, return 2500.
5. Otherwise, return `None`.

**Your turn:** For 5.1 km, which checks fail and which check returns a fee?

````python
def delivery_fee(distance_km):
    if distance_km < 0:
        raise ValueError("Distance cannot be negative.")
    if distance_km <= 5:
        return 1000
    if distance_km <= 10:
        return 1500
    if distance_km <= 20:
        return 2500
    return None
````

### 4 — Test
Run the next cell. Each `assert` checks a result and stops the cell if it is wrong. A check at a limit is called a **boundary test**.

````python
fee_cases = [(3, 1000), (5, 1000), (5.1, 1500), (10, 1500),
             (10.1, 2500), (20, 2500), (20.1, None)]
for distance, expected in fee_cases:
    actual = delivery_fee(distance)
    print(f"{distance} km | expected: {expected} | actual: {actual}")
    assert actual == expected

try:
    delivery_fee(-1)
except ValueError:
    print("Negative distance | expected: ValueError | actual: ValueError")
else:
    raise AssertionError("Negative distance was not rejected")
print("All delivery checks passed.")
````

### 5 — Reflect

## 4. Worked log: group articles by tag combination

**Request:** Group articles that have exactly the same set of tags, ignoring tag order and duplicate tags.

### 1 — Restate
Write `group_by_tag_combination(articles)`.

**Input:** A list of dictionaries. Each has a string `title` and a list of string `tags`.

**Output:** A dictionary:
- **Key:** A `frozenset` containing the group's distinct tags.
- **Value:** A list of article titles, in the order those articles appeared in the input.

**Rules:**
1. `["python", "web"]` and `["web", "python"]` belong together.
2. `["x", "x"]` and `["x"]` belong together.
3. `["x"]` and `["x", "y"]` do not belong together.
4. Empty input returns `{}`.
5. An empty tag list uses the key `frozenset()`.
6. Preserve repeated titles. Only duplicate tags are removed.

Assume inputs have the required structure. Do not change tag spelling, case, or spaces: `"Python"` and `"python"` are different tags.

### 2 — Examples
| Input: title → tags | Expected output |
|---|---|
| A → `["python", "web"]`; B → `["web", "python"]` | `{frozenset({"python", "web"}): ["A", "B"]}` |
| No articles | `{}` |
| A → `["x"]`; B → `["y"]` | `{frozenset({"x"}): ["A"], frozenset({"y"}): ["B"]}` |
| A → `["x", "x"]`; B → `["x"]` | `{frozenset({"x"}): ["A", "B"]}` |
| A → `[]`; B → `[]` | `{frozenset(): ["A", "B"]}` |

### Your turn — solve this by hand first
```python
[
    {"title": "Z", "tags": ["python", "web"]},
    {"title": "B", "tags": ["news"]},
    {"title": "A", "tags": ["web", "python", "python"]},
]
```
1. How many groups are needed?
2. What are their keys?
3. What title order belongs in each group?

Write your answer before reading the plan.

### 3 — Plan: turn the hand solution into steps

We need two things:
- One consistent key for equivalent tag lists.
- A place to collect titles under each key.

#### Choose the key
| Candidate | What goes wrong? |
|---|---|
| List of tags | A list cannot be a dictionary key. |
| Tuple of tags | Order and duplicate entries still matter. |
| Set of tags | Removes duplicates and ignores order, but cannot be a dictionary key. |
| **Frozenset of tags** | **Removes duplicates, ignores order, and can be a dictionary key.** |

A **frozenset** is a set that cannot be changed after creation. With string tags, it is valid as a dictionary key.

#### Write the actions
1. Create an empty dictionary called `groups`.
2. Visit each article in input order.
3. Convert its tags to a `frozenset` called `key`.
4. If `key` is new, store an empty list at `groups[key]`.
5. Append the article title to `groups[key]`.
6. Return `groups` after the loop.

#### Trace the actions on the example
| Article | Key | Action | Group's titles afterward |
|---|---|---|---|
| Z | `frozenset({"python", "web"})` | Create list, append Z | `["Z"]` |
| B | `frozenset({"news"})` | Create list, append B | `["B"]` |
| A | Same key as Z | Append A to existing list | `["Z", "A"]` |

This trace answers the three questions: **two groups**, one for python/web and one for news. Python/web contains `["Z", "A"]`; news contains `["B"]`.

````python
def group_by_tag_combination(articles):
    groups = {}

    for article in articles:
        key = frozenset(article["tags"])

        if key not in groups:
            groups[key] = []

        groups[key].append(article["title"])

    return groups

articles = [
    {"title": "Z", "tags": ["python", "web"]},
    {"title": "B", "tags": ["news"]},
    {"title": "A", "tags": ["web", "python", "python"]},
]
print(group_by_tag_combination(articles))
````

### 4 — Test
For each case below, compare actual output with the expected dictionary.

The printed order of tags inside a `frozenset` may vary. That is not a failure. The order of titles inside each list **must** match.

````python
tag_cases = [
    ("Tag order does not matter", [
        {"title": "A", "tags": ["python", "web"]},
        {"title": "B", "tags": ["web", "python"]},
    ], {frozenset({"python", "web"}): ["A", "B"]}),

    ("Empty articles", [], {}),

    ("Different combinations stay separate", [
        {"title": "A", "tags": ["x"]},
        {"title": "B", "tags": ["y"]},
    ], {frozenset({"x"}): ["A"], frozenset({"y"}): ["B"]}),

    ("Duplicate tags do not matter", [
        {"title": "A", "tags": ["x", "x"]},
        {"title": "B", "tags": ["x"]},
    ], {frozenset({"x"}): ["A", "B"]}),

    ("Empty tags share one group", [
        {"title": "A", "tags": []},
        {"title": "B", "tags": []},
    ], {frozenset(): ["A", "B"]}),

    ("An extra tag creates a different group", [
        {"title": "A", "tags": ["x"]},
        {"title": "B", "tags": ["x", "y"]},
    ], {frozenset({"x"}): ["A"], frozenset({"x", "y"}): ["B"]}),

    ("Titles keep input order across interleaved groups", [
        {"title": "Z", "tags": ["python", "web"]},
        {"title": "B", "tags": ["news"]},
        {"title": "A", "tags": ["web", "python"]},
        {"title": "D", "tags": ["news"]},
    ], {frozenset({"python", "web"}): ["Z", "A"],
        frozenset({"news"}): ["B", "D"]}),

    ("Repeated titles are kept", [
        {"title": "A", "tags": ["x"]},
        {"title": "A", "tags": ["x"]},
    ], {frozenset({"x"}): ["A", "A"]}),

    ("Case is not changed", [
        {"title": "A", "tags": ["Python"]},
        {"title": "B", "tags": ["python"]},
    ], {frozenset({"Python"}): ["A"], frozenset({"python"}): ["B"]}),
]

for name, test_articles, expected in tag_cases:
    actual = group_by_tag_combination(test_articles)
    assert actual == expected, f"{name}: expected {expected!r}, got {actual!r}"
    assert all(isinstance(key, frozenset) for key in actual)
    print("PASS:", name)

print(f"All {len(tag_cases)} grouping tests passed.")
````

### How to repair a failed test
Suppose you used `groups[key] = [article["title"]]` on every iteration instead of appending.

For two matching articles A then B:
1. After A, the list is `["A"]`.
2. After B, the list becomes `["B"]`.
3. Expected: `["A", "B"]`. Actual: `["B"]`.
4. The first wrong action is replacing the list when B arrives.
5. Fix: create a list only for a new key; append each title afterward.
6. Rerun all tests, not just the one that failed.

### 5 — Reflect
The tag-grouping tests pass. A `frozenset` matches the rule for tag equality; appending titles preserves their input order. Next time, I will decide what makes two items equal before choosing a dictionary key.

## 9. Worked log: check whether two texts are anagrams

### 1 — Restate
Write `are_anagrams(first, second)`.

Anagrams contain **the same letters the same number of times**, possibly in a different order. For this exercise:
- Inputs are strings containing English letters, spaces, or punctuation; digits are outside scope.
- Ignore uppercase/lowercase differences, spaces, and punctuation.
- Return `True` if the remaining letters match in both text and frequency; otherwise return `False`.
- Identical words count as a match. Two inputs with no letters also count as a match.

**Important:** Same distinct letters is not enough. `"abb"` and `"ab"` are not anagrams: one has an extra `b`.

### 2 — Examples
| First | Second | Expected | Why |
|---|---|---|---|
| `"listen"` | `"silent"` | True | Same letters and counts |
| `"Dormitory"` | `"Dirty room!!"` | True | Ignore case, spaces, punctuation |
| `"hello"` | `"world"` | False | Different letters |
| `"abb"` | `"ab"` | False | Different counts |
| `"aab"` | `"abb"` | False | Same length and distinct letters, different counts |
| `"cat"` | `"cat"` | True | Identical words satisfy the rule |
| `""` | `"!!!"` | True | Both have no letters after cleaning |

**Your turn:** Are `"rail safety"` and `"fairy tales"` anagrams? Count the letters before checking with code.

### 3 — Plan and solve
1. For each input, keep only letters and convert them to lowercase.
2. Sort the letters of each cleaned input.
3. Compare the two sorted lists. Return the result of that comparison.

**Why sorting works:** Two matching letter collections produce the same sorted list. Sorting keeps repeated letters; a set would discard them.

**Trace:**
- `"Listen!"` → `"listen"` → `['e', 'i', 'l', 'n', 's', 't']`
- `"Silent"` → `"silent"` → `['e', 'i', 'l', 'n', 's', 't']`
- Lists match → return `True`.

````python
def are_anagrams(first, second):
    first_letters = []
    for character in first:
        if character.isalpha():
            first_letters.append(character.lower())

    second_letters = []
    for character in second:
        if character.isalpha():
            second_letters.append(character.lower())

    return sorted(first_letters) == sorted(second_letters)

print("rail safety / fairy tales:", are_anagrams("rail safety", "fairy tales"))
````

### 4 — Test
Run every example, not just one successful pair. The count tests catch an incorrect solution that compares sets of letters.

````python
anagram_cases = [
    ("listen", "silent", True),
    ("Dormitory", "Dirty room!!", True),
    ("hello", "world", False),
    ("abb", "ab", False),
    ("aab", "abb", False),
    ("cat", "cat", True),
    ("", "!!!", True),
    ("", "a", False),
    ("rail safety", "fairy tales", True),
]
for first, second, expected in anagram_cases:
    actual = are_anagrams(first, second)
    print(f"{first!r} / {second!r} | expected: {expected} | actual: {actual}")
    assert actual == expected
print("All anagram checks passed.")
````

### 5 — Reflect
The tests confirm the agreed cleaning rules and letter counts for these examples. Sets would incorrectly treat `"aab"` and `"abb"` as equal; sorting preserves the counts. Next time, decide whether repetitions matter before choosing a data structure.

**Connect it to tag grouping:** Duplicate tags did not matter, so `frozenset` was suitable. Duplicate letters do matter here, so it is not suitable.

## 10. Worked log: check whether text is a palindrome

### 1 — Restate
Write `is_palindrome(text)`.

A palindrome reads the same **forward and backward**. For this exercise:
- Input is a string containing English letters, digits, spaces, or punctuation.
- Ignore case, spaces, and punctuation; keep letters and digits.
- Return `True` when the cleaned text equals its reverse; otherwise return `False`.
- Empty cleaned text and one-character text count as palindromes.

Unlike an anagram check, **position matters**. We compare characters at opposite ends, not sorted characters.

### 2 — Examples
| Input | Cleaned text | Expected |
|---|---|---|
| `"level"` | `"level"` | True |
| `"RaceCar"` | `"racecar"` | True |
| `"A man, a plan, a canal: Panama!"` | `"amanaplanacanalpanama"` | True |
| `"hello"` | `"hello"` | False |
| `"abca"` | `"abca"` | False |
| `"1221"` | `"1221"` | True |
| `"12"` | `"12"` | False |
| `"x"` | `"x"` | True |
| `"!!!"` | `""` | True |
| `""` | `""` | True |

**Your turn:** Is `"Never odd or even"` a palindrome? Remove the spaces and check opposite ends.

### 3 — Plan and solve
1. Start with an empty list of characters.
2. Visit each input character. If it is a letter or digit, append its lowercase form.
3. Compare the list with a reversed copy.
4. Return `True` if they match, otherwise return `False`.

**Trace for `"Level!"`:**
- Keep letters and digits, lowercase → `['l', 'e', 'v', 'e', 'l']`.
- Reverse → `['l', 'e', 'v', 'e', 'l']`.
- Compare → equal → `True`.

**Trace for `"abca"`:** The outside `a` characters match, but the inside `b` and `c` do not. The full list and reverse differ → `False`.

````python
def is_palindrome(text):
    characters = []
    for character in text:
        if character.isalnum():
            characters.append(character.lower())

    # [::-1] creates a reversed copy of the list.
    return characters == characters[::-1]

print("Never odd or even:", is_palindrome("Never odd or even"))
````

### 4 — Test
Include an even-length palindrome, an odd-length palindrome, and a word whose ends match but middle does not. Digits must be retained: `"12"` must return `False`, not become empty text.

````python
palindrome_cases = [
    ("level", True),
    ("RaceCar", True),
    ("A man, a plan, a canal: Panama!", True),
    ("hello", False),
    ("abca", False),
    ("1221", True),
    ("12", False),
    ("x", True),
    ("!!!", True),
    ("", True),
    ("Never odd or even", True),
]
for text, expected in palindrome_cases:
    actual = is_palindrome(text)
    print(f"{text!r} | expected: {expected} | actual: {actual}")
    assert actual == expected
print("All palindrome checks passed.")
````
