## 1. The problem Git solves

Imagine eight people working on the same Python project:

- Person A changes the main Python file.
- Person B changes that same file.
- Person C accidentally removes Person A's work.
- Nobody knows who changed what or when.
- Someone makes a mistake, and the team cannot easily identify the last working version.

How can a team coordinate changes without repeatedly sending files to one another?

**Git** helps solve this problem. It is a version-control system that tracks changes to files over time. It helps developers preserve project history, inspect changes and work on separate lines of development.

### Git vs GitHub

- **Git:** the version-control tool installed and used on your computer.
- **GitHub:** an online platform where Git repositories can be hosted, shared and reviewed.

They work together, but they are not the same thing.

### What is a repository?

A **repository** (or repo) is a project directory tracked by Git. It contains the project's files and the history of changes Git has recorded.

## 2. Commits: recording snapshots of your work

A **commit** is a recorded snapshot of staged changes in a repository's history.

Imagine that you create `main.py`, add a greeting function and commit it. Later, you add another function and create another commit. Git now has a history of those changes.

### The basic sequence

1. Edit or create files.
2. Check what changed with `git status`.
3. Stage the changes with `git add`.
4. Record them with `git commit`.
5. Inspect the history with `git log`.

**Important distinction:** `git add` does not create a commit. It selects changes to be included in the next commit.

````python
# Run these commands in a terminal, not in a Python cell.
# First, create a project folder and enter it:
mkdir git-learning-demo
cd git-learning-demo

# Initialise Git in this folder:
git init
````

````python
# Create a file called main.py, then stage and commit it:
git status
git add main.py
git commit -m "Add greeting function"

# Inspect the commit history:
git log --oneline
````

### Important commands

| Command | What it does |
|---|---|
| `git init` | Initialises a Git repository in the current directory. |
| `git status` | Shows the state of the working directory and staging area. |
| `git add <file>` | Stages selected changes. |
| `git add .` | Stages changes under the current directory. Check `git status` before committing. |
| `git commit -m "message"` | Records staged changes in the repository history. |
| `git log --oneline` | Shows a compact commit history. |

### What makes a good commit message?

A good commit message describes the change clearly and briefly.

- Better: `Add greeting function`
- Less useful: `Changes`
- Less useful: `Update stuff`

**Teaching check:** If someone sees your commit message three weeks later, should they be able to understand what changed?

## 3. Branches: working separately

A **branch** is a separate line of development. It lets a developer work on a feature or fix without directly changing another branch.

For example:

- `main` is commonly the primary branch.
- `feature/add-greeting` is a branch for a new greeting feature.

You can make commits on the feature branch while `main` remains unchanged by those commits. Switching back to `main` does not automatically bring the feature into it.

````python
# List branches:
git branch

# Create a branch and switch to it:
git switch -c feature/add-greeting

# After editing your file, stage and commit:
git add .
git commit -m "Add greeting feature"

# Switch back to the main branch:
git switch main

# List branches again:
git branch
````

### Branch commands

| Command | What it does |
|---|---|
| `git branch` | Lists local branches. |
| `git switch -c feature/name` | Creates a branch and switches to it. |
| `git switch main` | Switches to the `main` branch. |
| `git branch -d feature/name` | Deletes a local branch that Git considers safely merged. |

**Ask the class:** Why might a team prefer feature branches instead of asking everyone to make changes directly on `main`?

## 4. Merging: bringing changes together

**Merging** combines changes from one branch into another.

If a feature is ready to be incorporated into `main`, switch to `main` and merge the feature branch:

```bash
git switch main
git merge feature/add-greeting
```

Git may combine the changes automatically. Sometimes two branches make incompatible changes to the same part of a file. Git may then report a **merge conflict**.

A merge conflict is not necessarily a disaster. It means a person needs to inspect the conflicting changes, decide what the final content should be, and complete the merge.

## 5. Pull requests: proposing changes for review

A **pull request (PR)** is a proposal to merge changes from one branch into another, usually on a platform such as GitHub.

A PR gives teammates a place to:

- Understand what changed and why.
- Inspect the proposed changes.
- Ask questions and leave feedback.
- Request corrections.
- Approve the contribution before it is merged.

### A pull request is not the same as a merge

- **Pull request:** proposes the changes and enables review.
- **Merge:** incorporates the changes into the target branch.

### Typical team workflow

1. Clone the repository.
2. Create a feature branch.
3. Make changes.
4. Stage and commit those changes.
5. Push the branch to GitHub.
6. Open a pull request.
7. Review the changes and address feedback.
8. Merge the approved pull request.

````python
# Example workflow after you have cloned a shared repository:

# Create and switch to a feature branch:
git switch -c feature/add-description

# Stage and commit your changes:
git add README.md
git commit -m "Add project description"

# Push the branch to GitHub:
git push -u origin feature/add-description
````

After pushing, open the repository on GitHub and create a pull request:

- **Base branch:** the branch receiving the changes, often `main`.
- **Compare branch:** the branch containing your proposed changes.
- **Title:** a short description of the change.
- **Description:** what changed, why it changed and how it can be tested.
- **Files changed:** inspect the actual changes before requesting review.

The exact GitHub interface can change, but the collaboration idea remains the same.

## 6. Why small pull requests are easier to review

Compare two contributions:

- **PR A:** changes two files to add one focused feature.
- **PR B:** changes 45 files, adds three features and modifies unrelated code.

PR A is generally easier to understand and review.

Small, focused PRs:

- Are easier to understand.
- Make mistakes easier to locate.
- Keep unrelated changes out of a review.
- Make feedback more focused.
- Are easier to test.

A useful PR description answers three questions:

1. What changed?
2. Why was it changed?
3. How can someone test it?
