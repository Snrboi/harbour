# HTTP Status Codes & Error Responses
## A beginner-friendly teaching note

This notebook starts from the basics and builds up to designing helpful API errors. It is written for someone who is new to HTTP and wants to learn the topic well enough to explain it to others.

**No code needs to be run.** The examples are shown as text so you can focus on the ideas.

## How to use this notebook

Read it in order the first time. As you go, try to explain each idea out loud in your own words. At the end there is a short practice section, a suggested way to teach the topic, and a one-page quick reference.

### By the end, you should be able to

- Explain what an HTTP status code tells us.
- Recognize the main groups: `1xx`, `2xx`, `3xx`, `4xx`, and `5xx`.
- Choose common error codes such as `400`, `401`, `403`, `404`, `409`, `422`, and `500`.
- Explain why an error response needs a useful, safe body as well as a status code.
- Teach the basic ideas using everyday examples.

## 1. What is HTTP?

**HTTP** is a set of rules that lets a client and a server communicate. A client might be a web browser, a phone app, or another program. A server receives the request, does some work, and sends a response.

Think of ordering food:

- You (the client) ask for something.
- The restaurant (the server) receives the request.
- The restaurant either completes it or tells you why it could not.

On the web, a request might ask for a page, a user profile, a book, or an action such as creating an account. The server's response includes a **status code**: a short signal about what happened.

A status code is not the whole explanation. It is more like a signpost: “success,” “your request needs attention,” or “the server had a problem.”

## 2. What is in a request and a response?

A simple HTTP request contains:

- **Method:** what the client wants to do, such as `GET` (read) or `POST` (send/create).
- **Path:** which resource it is about, such as `/books/42`.
- **Headers:** extra information, such as the content type.
- **Body:** optional data sent with the request, often JSON.

A response contains:

- **Status code:** the result, such as `200` or `404`.
- **Headers:** extra information about the response.
- **Body:** optional content, such as the requested data or an explanation of an error.

Example request:

```http
GET /books/42 HTTP/1.1
Accept: application/json
```

Example successful response:

```http
HTTP/1.1 200 OK
Content-Type: application/json

{"id": 42, "title": "The River Between"}
```

The first response line says the request succeeded. The JSON body contains the book data.

## 3. The five status-code groups

The first digit gives a quick clue about the kind of result:

| First digit | Group | Beginner's meaning |
|---|---|---|
| `1xx` | Informational | The request is still being handled or more communication is needed. |
| `2xx` | Success | The request worked, or was accepted. |
| `3xx` | Redirection | The client may need to use another location or a cached version. |
| `4xx` | Client/request problem | Something about this request, permission, or requested resource prevents success. |
| `5xx` | Server problem | The server or a service it depends on failed to complete a request that may otherwise be valid. |

For everyday API work, you will most often discuss `2xx`, `4xx`, and `5xx`. You do not need to memorize every code to understand the main idea.

## 4. Common success codes: `2xx`

| Code | Name | When it is commonly used |
|---|---|---|
| **200 OK** | The request succeeded. | Reading data or completing an action that returns a response body. |
| **201 Created** | A new resource was created. | Creating an account, book, order, or other resource. The response often includes the new resource or its location. |
| **202 Accepted** | The server accepted the request, but work is not finished yet. | A long-running job has been queued and will complete later. “Accepted” does not mean “finished.” |
| **204 No Content** | The request succeeded, and there is no response body. | A successful delete or update where the API has nothing else to return. |

A useful teaching point: `200` is not the only success code. Choose the success code that describes what actually happened.

## 5. Common client/request errors: `4xx`

A `4xx` code says the server could not fulfil the request as sent. The exact cause depends on the code.

| Code | Name | Plain-language meaning | Example |
|---|---|---|---|
| **400** | Bad Request | The request is malformed or cannot be understood. | The client sends broken JSON. |
| **401** | Unauthorized | The client is not properly authenticated; it may need to sign in or provide valid credentials. | A private endpoint is called without a valid login token. |
| **403** | Forbidden | The server understands who is asking, but that user is not allowed to do this. | A regular user tries to open an administrator-only page. |
| **404** | Not Found | The requested resource cannot be found, or the server intentionally hides it. | The client asks for a profile ID that does not exist. |
| **405** | Method Not Allowed | The path exists, but that HTTP method is not allowed there. | The client tries `DELETE` on a read-only endpoint. |
| **409** | Conflict | The request conflicts with the current state of the resource. | The client tries to create a second active booking for the same room and time. |
| **415** | Unsupported Media Type | The server does not accept the format the client sent. | The client sends XML to an endpoint that only accepts JSON. |
| **422** | Unprocessable Content | The server can read the request, but the content fails validation or a rule. | The client sends a valid JSON number for `age`, but the value is below the allowed minimum. |
| **429** | Too Many Requests | The client has sent requests too quickly or exceeded a limit. | A service limits a user to 100 requests per minute. |

The exact use of some codes—especially `400` and `422`—can vary between APIs. A well-designed API chooses a convention, documents it, and applies it consistently.

## 6. Common server errors: `5xx`

A `5xx` code means the server could not complete the request because of a server-side or upstream problem.

| Code | Name | Plain-language meaning | Example |
|---|---|---|---|
| **500** | Internal Server Error | Something unexpected went wrong in the server. | A software bug causes the request to fail. |
| **502** | Bad Gateway | A server acting as a gateway received a bad response from another server. | An API gateway cannot get a valid response from an upstream service. |
| **503** | Service Unavailable | The service is temporarily unable to handle requests. | The service is down for maintenance or temporarily overloaded. |
| **504** | Gateway Timeout | A gateway did not receive a timely response from another server. | A search service takes too long to answer through the API gateway. |

A `500` is not a way to describe every problem. If the service is temporarily unavailable, `503` may describe the situation better. The API team should choose the code that best matches what happened.

## 7. Four pairs people often mix up

### `400` versus `422`

- **400:** the server cannot properly understand the request as sent. For example, the JSON is broken.
- **422:** the request can be read, but its content breaks a validation rule. For example, the JSON is valid but a required field is missing or has an unacceptable value.

Some APIs use `400` for more kinds of input errors. That can be fine if the API documents the rule and is consistent.

### `401` versus `403`

- **401:** “I cannot verify who you are; sign in or send valid credentials.”
- **403:** “I know who you are, but you do not have permission for this action.”

### `404` versus `409`

- **404:** the requested resource is not found (or is deliberately hidden).
- **409:** the resource exists, but its current state conflicts with the requested action.

### `500` versus `503`

- **500:** an unexpected server failure occurred.
- **503:** the service is temporarily unable to serve the request.

**Memory aid:** “not found” is about whether the resource is there; “conflict” is about whether the requested action fits the resource's current state.

## 8. An error response has two jobs

A good error response has:

1. A **status code** that lets software identify the broad kind of outcome.
2. A **response body** that gives a safe, useful explanation.

For example, `404` tells a program “not found.” A body can tell the person using the program “We could not find a profile with that ID.” It can also include a stable error code so the app can choose what to do without reading a sentence.

A response body is not always required. For example, `204 No Content` intentionally has no body. But when an API returns an error body, it should be clear and consistent.

## 9. Example of a helpful error response

This example uses **Problem Details**, a standard format for describing HTTP API errors. The body is JSON; `application/problem+json` is the content type commonly used for this format.

```http
HTTP/1.1 422 Unprocessable Content
Content-Type: application/problem+json

{
  "type": "https://api.example.com/problems/validation-error",
  "title": "The request has invalid fields",
  "status": 422,
  "detail": "Correct the fields listed in errors and try again.",
  "errors": [
    {
      "field": "email",
      "code": "invalid_format",
      "message": "Enter an email address in the expected format."
    }
  ],
  "request_id": "req-7f31"
}
```

What each part is for:

- `type`: a stable identifier for this kind of problem.
- `title`: a short summary.
- `status`: repeats the HTTP status so a reader can understand the body by itself. It should match the actual HTTP status.
- `detail`: a human-readable explanation.
- `errors`: optional field-by-field help.
- `request_id`: helps support staff find the matching server log entry.

An API may use slightly different field names or add fields. The key ideas are consistency, clarity, and safety.

## 10. How to write good error messages

A helpful message answers these questions:

- **What happened?** “The email address is not in an accepted format.”
- **What can the user do next?** “Enter a valid email address and try again.”
- **Can the response be safely shown to this user?** Do not reveal another person's information or private system details.

### Good habits

- Use plain language and be specific enough to help.
- Keep the structure consistent across the API.
- Include a stable machine-readable code, such as `invalid_email` or `booking_conflict`.
- For a validation error, identify the field and the rule it broke.
- Include a request or trace ID when it will help support investigate.
- Make sure the HTTP status and the body's `status` field agree.

### Avoid

- Returning a database query, password, stack trace, file path, or internal server details to the user.
- Making software depend on exact English wording. Wording may change or be translated.
- Sending `200 OK` for an unsuccessful request just because the body contains an `"error"` field.

A useful balance: **give the user enough information to fix the problem, but not information they should not see.**

## 11. A simple way to choose a status code

When designing or reading an API response, ask:

1. **Did the request succeed?** Use an appropriate `2xx` code.
2. **Is the request malformed or impossible to interpret?** Consider `400`.
3. **Is the user signed in?** Missing or invalid authentication may mean `401`.
4. **Is the user allowed to do this?** If not, consider `403`.
5. **Does the requested resource exist or appear to the caller?** If not, consider `404`.
6. **Does the request conflict with the resource's current state?** Consider `409`.
7. **Can the request be read, but one or more values fail validation?** Consider `422` under the API's convention.
8. **Is the server temporarily unavailable or overloaded?** Consider `503`; a gateway timeout may be `504`.
9. **Did an unexpected server failure occur?** Consider `500`.

This is a guide, not a substitute for an API's documentation. The same API should make the same decision for the same kind of problem.

## 12. Worked examples

### Example A: broken request format

A phone app sends text that is not valid JSON. The server cannot read the request body. **A common choice is `400 Bad Request`.** The response can say that the body must be valid JSON. The user should correct the request; repeating it unchanged will not help.

### Example B: a missing page or item

A user asks for `/products/123`, but product `123` does not exist. **`404 Not Found`** tells the app the resource could not be found.

### Example C: a user lacks permission

A user is signed in but is not allowed to view an administrator report. **`403 Forbidden`** is a common choice. In some security-sensitive systems, an API intentionally uses `404` so outsiders cannot tell that the report exists. That should be a deliberate, consistent security decision.

### Example D: a booking conflict

A room already has a booking for 2 p.m., and another request tries to book the same room at the same time. **`409 Conflict`** explains that the request conflicts with the current state. The user may choose another time.

### Example E: a service is temporarily down

The API is temporarily unavailable during maintenance. **`503 Service Unavailable`** may be appropriate. The server may include a `Retry-After` header to suggest when the client should try again.

### Example F: too many requests

A client exceeds a usage limit. **`429 Too Many Requests`** can tell it to slow down. A `Retry-After` header may say how long to wait.

## 13. Should the client try again?

Not every error should be retried.

- **Usually fix the request first:** `400`, `401`, `403`, `404`, and `422`. Repeating exactly the same request is unlikely to help.
- **Check the situation:** `409`. The client may need to refresh the resource or choose a different action.
- **Slow down:** `429`. Follow `Retry-After` if present.
- **Possibly retry later:** `503` or `504`, depending on the API and the operation.
- **Use caution with `500`:** a temporary retry might work, but the cause is unknown.

A **GET** request is normally intended to be safe to repeat. An action such as charging a card or creating an order may have side effects. If a client repeats it after a timeout, the server might perform it twice. APIs can use **idempotency keys** to recognize a retry of the same operation. The lesson is: do not tell people “always retry 5xx” without considering whether repeating the action is safe.

## 14. Common misunderstandings

- **“A 404 means the whole website is down.”** Usually it means this particular path or resource was not found.
- **“Every successful request returns 200.”** A new resource may return 201; accepted background work may return 202; a successful response with no body may return 204.
- **“401 and 403 mean the same thing.”** In simple terms, 401 is about authentication; 403 is about permission.
- **“The status code explains everything.”** It gives a broad signal; a safe response body can explain the specific problem.
- **“Every invalid input must be 400.”** APIs differ; 400 and 422 usage should be consistent and documented.
- **“A 500 response should show the programmer the error.”** Detailed diagnostics belong in protected logs, not in a response visible to an untrusted caller.
- **“The message text is an API contract.”** Client code should use documented status codes and stable error codes, not rely on exact wording.

### Practice answers

The point is not to memorize answers without context. Practice explaining **what happened** and **why that status fits**.

## 16. A simple plan for teaching this topic

You can teach the basics in about 10–15 minutes, adjusting to your group's time.

### 1. Start with the client/server idea

Say: “A client asks a server to do something. The server replies with a status code and possibly a body.” Use a browser loading a page as your familiar example.

### 2. Explain the first digit

Show the groups: `2xx` success, `3xx` redirect, `4xx` request/client problem, `5xx` server problem. Explain that `1xx` is informational and less common for beginners.

### 3. Teach a small set of useful codes

Start with `200`, `201`, `400`, `401`, `403`, `404`, `409`, `422`, and `500`. Add `429` and `503` as examples of rate limiting and temporary outage.

### 4. Compare the pairs

Ask learners to explain `401` versus `403`, `404` versus `409`, and `400` versus `422`. Use one short example for each.

### 5. Show the error response body

Explain that the status is the broad signal; the body gives safe detail. Point out a human-readable message, a stable error code, and a request ID.

### 6. Let the learners decide

Read the eight practice scenarios without showing the answers. Invite a reason, not only a number. Correct misunderstandings kindly.

### 7. Finish with the main rule

> “Choose a status code that matches what happened. Give a consistent, safe explanation in the body. Document the API's choices so clients know what to do.”

## 17. Quick-reference sheet

### Success

- `200` — succeeded, often with a body.
- `201` — created a resource.
- `202` — accepted for later processing.
- `204` — succeeded, no body.

### Request, access, or resource problem

- `400` — malformed or otherwise bad request.
- `401` — authentication is missing or invalid.
- `403` — authenticated/known caller is not allowed.
- `404` — resource not found or deliberately hidden.
- `405` — method not allowed on this path.
- `409` — conflict with current state.
- `415` — unsupported request format.
- `422` — parseable request fails validation, if that is the API's convention.
- `429` — too many requests.

### Server problem

- `500` — unexpected server failure.
- `502` — bad response from an upstream server.
- `503` — temporarily unavailable.
- `504` — upstream response timed out.

### Error-body reminders

- Match the body status to the real HTTP status.
- Explain the problem in safe, plain language.
- Provide a stable machine-readable code where useful.
- Do not expose secrets or stack traces.
- Keep the format consistent.

## 18. Glossary and further reading

- **Client:** the browser, app, or program making a request.
- **Server:** the program receiving the request and sending a response.
- **Endpoint:** a particular API path and method, such as `GET /books`.
- **Status code:** the three-digit result of an HTTP request.
- **Header:** metadata sent with a request or response.
- **Body:** the main content sent with a request or response.
- **Authentication:** checking who the user or program is.
- **Authorization:** checking what that user or program is allowed to do.
- **Validation:** checking that supplied data has the expected shape and values.
- **Idempotent operation:** an operation that can be repeated without changing the final result after the first successful attempt.

For deeper reading, look up **RFC 9110: HTTP Semantics** and **RFC 9457: Problem Details for HTTP APIs**. You do not need to memorize the RFCs to teach the beginner concepts in this notebook.

---

### Final recap

A status code is a short, standardized signal about an HTTP request. The first digit tells you the broad category. Choose a specific code that matches what happened, then use a consistent, safe error body to explain the details. When teaching, focus on examples and contrasts—not memorizing a long list.
