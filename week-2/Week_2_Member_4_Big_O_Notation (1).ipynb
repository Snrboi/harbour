## 2. What does Big-O mean?

Let `n` represent the size of the input. Big-O describes how the work performed by an algorithm grows as `n` grows.

For example:
- A fixed number of operations is **O(1)**.
- One pass through `n` items is **O(n)**.
- A loop inside another loop over `n` items is often **O(n²)**.

We usually ignore constant factors and lower-order terms when describing asymptotic growth.

Example: `3n² + 5n + 10` is **O(n²)** because the quadratic term dominates for large `n`.

**Important distinction:** Big-O is not a stopwatch measurement. Two O(n) implementations can have different actual runtimes because of constants, hardware, language implementation, memory access and input distribution.

````python
def show_linear_work(items):
    for item in items:
        print(item)

show_linear_work(["A", "B", "C"])
````

The loop body runs once for each item. With `n` items, the number of iterations grows proportionally to `n`: **O(n)** time.

## 3. Common complexity classes

| Complexity | Informal name | Typical example |
|---|---|---|
| O(1) | Constant | Access a list element by index |
| O(log n) | Logarithmic | Binary search in sorted data |
| O(n) | Linear | Scan a list once |
| O(n log n) | Linearithmic | Typical efficient comparison sorting |
| O(n²) | Quadratic | Compare every pair using nested loops |
| O(2ⁿ) | Exponential | A naive exhaustive subset search |

The table gives common patterns, not guarantees for every implementation. For example, sorting algorithms have different best, average and worst cases.

````python
# O(1): direct indexing
books = ["Dune", "Kindred", "Beloved", "Things Fall Apart"]
selected_book = books[2]
print(selected_book)
````

**Explain:** Getting an element by index does not require scanning every element. For Python lists, indexing is O(1).

## 4. Linear time — O(n)

A single pass over a collection usually performs work proportional to the number of elements.

````python
def count_overdue(loans):
    count = 0
    for loan in loans:
        if loan["overdue"]:
            count += 1
    return count

sample_loans = [
    {"loan_id": "L1", "overdue": True},
    {"loan_id": "L2", "overdue": False},
    {"loan_id": "L3", "overdue": True},
]
count_overdue(sample_loans)
````

There is one loop over the loans, so the time complexity is **O(n)**. The counter uses a fixed amount of extra memory, so auxiliary space is **O(1)**.

**Ask the team:** If the number of loans doubles, what happens to the number of loop iterations? It approximately doubles too.

## 5. Logarithmic time — O(log n)

An algorithm is logarithmic when each step reduces the remaining problem by a constant factor, often by half.

Binary search works on sorted data:
1. Inspect the middle item.
2. Discard the half that cannot contain the target.
3. Repeat on the remaining half.

Because the search range is halved each time, the number of steps grows roughly with `log₂(n)`.

````python
def binary_search(sorted_ids, target):
    low, high = 0, len(sorted_ids) - 1

    while low <= high:
        middle = (low + high) // 2
        if sorted_ids[middle] == target:
            return middle
        if sorted_ids[middle] < target:
            low = middle + 1
        else:
            high = middle - 1

    return -1

ids = ["L001", "L003", "L007", "L010", "L014", "L020"]
binary_search(ids, "L010")
````

**Complexity:** O(log n) time and O(1) auxiliary space for this iterative version.

**Trade-off:** Binary search is fast, but the input must be sorted. Sorting first has a cost; if the data changes frequently, maintaining sorted order may also be expensive.

## 6. Quadratic time — O(n²)

A nested loop that compares each item with many other items often produces quadratic work.

````python
def print_all_pairs(book_ids):
    for first in book_ids:
        for second in book_ids:
            print(first, second)

print_all_pairs(["B1", "B2", "B3"])
````

For `n` IDs, the outer loop runs `n` times and the inner loop runs `n` times for each outer iteration: `n × n = n²` operations.

If the input doubles, the pairwise work grows by about four times.

Be precise: a nested loop is not automatically O(n²). If the inner loop runs a fixed number of times, or the total work is bounded differently, analyse the actual iteration counts.

## 7. Consecutive loops versus nested loops

Two separate loops over the same input perform approximately `n + n = 2n` work. Drop the constant factor: **O(n)**.

A loop inside another loop can perform `n × n = n²` work: **O(n²)**.

The key is to count total work, not simply count how many `for` keywords appear.

````python
def two_passes(items):
    # First pass: O(n)
    for item in items:
        _ = item.lower()

    # Second pass: O(n)
    for item in items:
        _ = len(item)

# Total: O(n) + O(n) = O(2n) = O(n)
two_passes(["Dune", "Beloved", "Kindred"])
````

## 8. Data structures change the cost

Choosing the right data structure can change an algorithm's growth rate.

| Operation | Python list | Python set/dict |
|---|---:|---:|
| Membership / key lookup | O(n) | O(1) average |
| Add an item | append O(1) amortized | O(1) average |
| Remove/search by value | O(n) | O(1) average by key/value membership for set |
| Access list item by index | O(1) | Not applicable |

These are typical/average-case costs. Hash-table operations can degrade in pathological collision cases; the average-case assumption should be stated.

**Memory trade-off:** Sets and dictionaries generally use more memory than a compact list because they maintain hash-table structures. In return, they offer fast average-case membership and key lookup.

````python
book_ids = ["B001", "B002", "B003", "B004"]

# List membership: O(n) worst case
"B004" in book_ids

# Set membership: O(1) average case
book_id_set = set(book_ids)
"B004" in book_id_set
````

## 9. Worked example: detect duplicate loan IDs in one pass

This is your Friday mini-project task from the README.

### Approach A: compare each ID with every other ID

A pairwise comparison approach can take O(n²) time. It is straightforward, but it becomes expensive as the list grows.

### Approach B: use a dictionary/set to remember IDs already seen

Scan once. For each ID:
- If it is already in `seen`, it is a duplicate.
- Otherwise, add it to `seen`.

Expected time: **O(n)** average-case.
Extra space: **O(n)** because the set may store up to `n` IDs.

This is a classic time-memory trade-off: use extra memory to avoid repeated scanning.

````python
def find_duplicate_loan_ids(loan_ids):
    seen = set()
    duplicates = set()

    for loan_id in loan_ids:
        if loan_id in seen:
            duplicates.add(loan_id)
        else:
            seen.add(loan_id)

    return duplicates

loan_ids = ["LN-101", "LN-102", "LN-103", "LN-102", "LN-104", "LN-101"]
find_duplicate_loan_ids(loan_ids)
````

### Walk through the example

| Step | Current ID | Already seen? | Action |
|---:|---|---|---|
| 1 | LN-101 | No | Add to `seen` |
| 2 | LN-102 | No | Add to `seen` |
| 3 | LN-103 | No | Add to `seen` |
| 4 | LN-102 | Yes | Record duplicate |
| 5 | LN-104 | No | Add to `seen` |
| 6 | LN-101 | Yes | Record duplicate |

**Why is this one pass?** The input list is traversed once. Set membership and insertion are O(1) on average, so total expected time is O(n).

**Why not claim guaranteed O(n)?** Hash-table operations are average-case O(1), not unconditional worst-case O(1). State the assumption when explaining the result.

## 10. A practical analysis checklist

When someone gives you a function, ask:

1. What is the input size `n`?
2. Which operations repeat as `n` grows?
3. Are loops sequential or nested?
4. Does a loop halve the search space?
5. Which data structures are used, and what are their operation costs?
6. What is the time complexity?
7. What extra memory is used?
8. What assumptions or trade-offs matter?

### Common mistakes to avoid
- Treating Big-O as exact runtime in seconds.
- Saying every nested loop is automatically O(n²).
- Forgetting that list membership is O(n).
- Describing dictionary lookup as guaranteed O(1), rather than average O(1).
- Ignoring extra memory when using a set or dictionary.
- Ignoring input preconditions, such as binary search requiring sorted input.

## 12. Closing summary

- Big-O describes how resource use grows with input size.
- A single scan is commonly O(n); nested full scans are commonly O(n²).
- Repeated halving produces O(log n).
- Data structures matter: a set/dictionary can replace repeated linear membership checks with average constant-time lookups.
- The duplicate-loan-ID solution uses a set to achieve expected O(n) time at O(n) extra space.

**Your key takeaway:** Always justify complexity by counting operations and explaining the data structures—not by guessing from the code's appearance.
